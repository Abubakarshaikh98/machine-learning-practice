## K-means Clustering
- K-mean clustering is a type of unsupervised learning algorithm.
- It works on unlabeled data.
- It make clusters from unlabeled data

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
import matplotlib.pyplot as plt
```

```python
iris = load_iris()
```

```python
x = pd.DataFrame(iris.data, columns=iris.feature_names)
```

```python
k = 3
```

```python
from sklearn.cluster import KMeans
kmeans = KMeans(n_clusters=k)
```

```python
kmeans.fit_predict(x)
```

```python
plt.scatter(x.iloc[:, 0], x.iloc[:, 1], c=kmeans.labels_, cmap='rainbow')
```

```python
plt.scatter(iris["data"][:, 0], iris["data"][:, 1], c=iris["target"], cmap='rainbow')
```

```python
df = pd.read_csv("calories.csv")
df.head()
```

```python
k_values = range(1, 11)
```

```python
x = df.iloc[:, 1:3]
```

# Different K values test karna

```python
inertia = []
```

```python
for k in k_values:
    kmeans = KMeans(n_clusters=k)
    kmeans.fit(x)
    inertia.append(kmeans.inertia_)
```

```python
plt.plot(range(1, 11), inertia, marker="o")

plt.xlabel("Number of Clusters (K)")
plt.ylabel("Inertia")
plt.title("Elbow Method")

plt.show()
```
